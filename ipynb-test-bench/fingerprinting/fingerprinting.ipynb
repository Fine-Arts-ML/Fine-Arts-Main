```python
from PIL import Image
import pandas as pd
import imagehash
import os
```

```python
img_list = []
for (root,dir,files) in os.walk('testdata/'):
    for file in files:
        img_list.append('testdata/' + str(file))

img_list
```

```python
hash_list =[]
for img in img_list:
    hash = imagehash.average_hash(Image.open(img))
    print(hash)
    hash_list.append([img, str(hash)])

hash_list
```
