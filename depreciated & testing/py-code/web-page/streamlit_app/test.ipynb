```python
def create_db_connection():
    load_dotenv()
    DB_HOST = os.getenv("DB_HOST")
    DB_NAME = os.getenv("DB_NAME")
    DB_USER = os.getenv("DB_USER")
    DB_PASSWORD = os.getenv("DB_PASSWORD")
    engine = create_engine('postgresql+pg8000://'+DB_USER+':'+DB_PASSWORD+'@'+DB_HOST+':5432/'+DB_NAME)
    return engine
```

```python

engine = create_db_connection()
metadata = MetaData()

# Reflect the table
view_avail_colors = Table('avail_colors', metadata, autoload_with=engine)

# Build the query
query = select(
    view_avail_colors.c.id,
    view_avail_colors.c.color_name
)

# Execute the query and fetch results
with engine.connect() as connection:
    result = connection.execute(query)

df_color_names = pd.DataFrame(result)

```
