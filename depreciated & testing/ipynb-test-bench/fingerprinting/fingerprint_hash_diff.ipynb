```python
from PIL import Image
import imagehash
import os
import pybktree
import collections
```

```python
img_list = []
for (root,dir,files) in os.walk('testdata/'):
    print(root)
    for file in files:
        if file not in os.walk('testdata/testimg/').__next__()[2]: # ignore input images
            img_list.append('testdata/' + str(file))

img_list
```

```python
img_hash_list=[]
for path in img_list:
    img = Image.open(path)
    img = img.resize((540,540))
    whash = imagehash.whash(img)
    ahash = imagehash.average_hash(img)
    phash = imagehash.phash(img)
    img_hash_list.append({"img": path,"w_hash": whash,"a_hash": ahash,"p_hash":phash})
    
```

```python
input_img_list = []
for (root,dir,files) in os.walk('testdata/testimg/'):
    for file in files:
        input_img_list.append('testdata/testimg/' + str(file))
input_img_list

input_hash_list=[]
for path in input_img_list:
    img = Image.open(path)
    img = img.resize((540,540))
    whash = imagehash.whash(img)
    ahash = imagehash.average_hash(img)
    phash = imagehash.phash(img)
    input_hash_list.append({"img": path,"w_hash": whash,"a_hash": ahash,"p_hash":phash})
```

```python
comparison_whash_list=[]
for item in input_hash_list:
    input_item = item['w_hash']
    input_name = item['img']
    print(input_name)
    tmp_list = []
    for source in img_hash_list:
        source_item = source['w_hash']
        source_name = source['img']
        diff = (input_item - source_item)/len(input_item.hash)**2
        tmp_list.append({"diff_whash":str(diff), 'filename':source_name})
        #print(f'''{(input_item - source_item)/len(input_item.hash)**2} filename:{source_name} ''')
    comparison_whash_list.append({'input':input_name ,"diff_whash_dict":tmp_list })
```

```python
for dict in comparison_whash_list:
    print(f'''input: {dict['input']} | matched file: { min(dict['diff_whash_dict'], key=lambda x: x['diff_whash'])}''')
```

```python
comparison_phash_list=[]
for item in input_hash_list:
    input_item = item['p_hash']
    input_name = item['img']
    tmp_list = []
    for source in img_hash_list:
        source_item = source['p_hash']
        source_name = source['img']
        diff = (input_item - source_item)/len(input_item.hash)**2
        tmp_list.append({"diff_phash":str(diff), 'filename':source_name})
    comparison_phash_list.append({'input':input_name ,"diff_phash_dict":tmp_list })

for dict in comparison_phash_list:
    print(f'''input: {dict['input']} | matched file: { min(dict['diff_phash_dict'], key=lambda x: x['diff_phash'])}''')
```
